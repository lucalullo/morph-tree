Replace file/target names with your data. No manual sampling or preprocessing is
required. Running fit is training and needs an appropriate finite resource scope.


```python
import pandas as pd
from morphtree import MorphTreeClassifier, MorphTreeRegressor

train = pd.read_csv("train.csv")
X = train.drop(columns=["target"])
y = train["target"]
X_test = pd.read_csv("test.csv")
```

```python
model = MorphTreeClassifier(random_state=42)
# For regression use MorphTreeRegressor(random_state=42).
model.fit(X, y)
predictions = model.predict(X_test)
```

```python
checksum = model.save("model.json", variant="all")
restored = MorphTreeClassifier.load("model.json", sha256=checksum)
assert restored.predict(X_test) == predictions
model.export_standalone("standalone.py")
model.export_algorithm("frozen_training.py")
model.save_recipe("recipe.json")
```
