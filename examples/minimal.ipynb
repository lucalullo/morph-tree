# MorphTree 2.1.0: classification and regression
Install MorphTree with `pip install morphtree`; install pandas separately to read CSV files.
Use a compatible environment as described in docs/INSTALLATION.md.
Replace file and target names with your data, keeping test columns aligned with the training features.
MorphTree handles supported missing and categorical values and plans training resources automatically.


```python
import pandas as pd
from morphtree import MorphTreeClassifier, MorphTreeRegressor

train = pd.read_csv("train.csv")
X = train.drop(columns=["target"])
y = train["target"]
X_test = pd.read_csv("test.csv")
```

```python
model = MorphTreeClassifier(random_state=42)
# For regression use MorphTreeRegressor(random_state=42).
model.fit(X, y)
predictions = model.predict(X_test)
if isinstance(model, MorphTreeClassifier):
    probabilities = model.predict_proba(X_test)
    second_class_probability = probabilities[:, 1]  # classes_[1]
```

```python
checksum = model.save("model.json", variant="all")
restored = type(model).load("model.json", sha256=checksum)
assert (restored.predict(X_test) == predictions).all()
model.export_standalone("standalone.py")
model.export_algorithm("frozen_training.py")
model.save_recipe("recipe.json")
```
